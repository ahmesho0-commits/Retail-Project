| Name | Description |
|------|-------------|
| Transaction_ID | Unique ID of one sales line |
| Order_ID | Order/basket ID; one order can have several rows |
| Customer_ID | Customer identifier (use this instead of the name) |
| Customer_Name | Customer full name; not a reliable key |
| Customer_Age | Customer age in years |
| Customer_Gender | Female / Male / Non-binary |
| Customer_Segment | Consumer, Small Business, Corporate, Premium, New Customer, Returning Customer |
| Order_Date | Date the order was placed |
| Order_Time | Time of day the order was placed |
| Sales_Channel | Online, Retail Store, B2B Portal, Phone Order |
| Store_ID | Store code |
| Store_Name | Store name (1-to-1 with Store_ID) |
| Country | Country of the store |
| Region | State / province / region |
| City | City of the store |
| Product_ID | Product code (28 products) |
| Product_Name | Product name (1-to-1 with Product_ID) |
| Product_Category | Electronics, Office Supplies, Furniture, Appliances |
| Product_Subcategory | Detailed group inside the category |
| Quantity | Units sold on the line |
| Unit_Price | List price per unit before discount |
| Discount_Percentage | Discount applied, in % |
| Sales_Amount | Revenue = Quantity × Unit_Price × (1 − Discount/100) |
| Cost_Amount | Total cost of goods for the line |
| Profit | Sales_Amount − Cost_Amount (negative = loss) |
| Payment_Method | Credit Card, Debit Card, PayPal, Bank Transfer, Apple Pay, Cash |
| Order_Status | Completed, Shipped, Processing, Returned, Cancelled |
| Shipping_Method | Standard, Express, Next Day, Pickup |
| Delivery_Days | Days from order to delivery (0 = same day / pickup) |
| Return_Flag | Yes / No: was the item returned |
| Return_Reason | Why it was returned; empty when Return_Flag = No |
| Sales_Representative | Sales rep who handled the order |
| Promotion_Code | Promo code used; empty = no promotion |
| Customer_Rating | Satisfaction score, 1 (worst) to 5 (best) |
| Inventory_Level | Stock units on hand at the time of sale |
| Order_Year | Year taken from Order_Date (2022–2025) |

### Check on a different Data frame to replace -ve values

```python
df_Quantity = df[df['Quantity']<1]
df_Quantity
```

```python
def Real_Quantity (row):
    if row['Quantity'] < 1:
        return (row['Sales_Amount'] // ((1 - row['Discount_Percentage'] / 100) * row['Unit_Price']))
    else:
        return row['Quantity']  

df_Quantity['Quantity'] = df_Quantity.apply(Real_Quantity, axis=1)

```

```python
df_Quantity['Quantity']
df_Quantity
```

```python
df['Quantity']= df.apply(Real_Quantity,axis=1)
```

```python
df['Quantity'].value_counts()
```

### Applied the function to Main df all Quantities are changed to logic numbers except a number is zero will be replaced to 1

```python
df ['Quantity'].replace({0 : 1 }, inplace= True)
```
